```python
## Feature Categories

'''- **ID**: transaction_id
- **Monetary**: amount_usd, account_balance_usd
- **Merchant/Card**: merchant_category, card_type, card_age_months, is_new_merchant, merchant_risk_score
- **Channel/Auth**: channel, device_type, auth_method, used_vpn
- **Behavioral**: hours_since_last_txn, txn_count_last_24h, velocity_score, cvv_retry_count, prior_disputes
- **Risk flags**: is_foreign_transaction, ip_country_mismatch, billing_shipping_mismatch, is_ai_generated_scam_attempt
- **Time**: time_of_day_hour, day_of_week
- **Demographic**: customer_age
- **Geo**: distance_from_home_km
- **Target**: is_fraud'''
```

```python
## Initial Observations

'''- Dataset: 20,000 transactions, 26 columns
- No missing values, no duplicate rows
- Target: is_fraud (binary) — ~1.7% fraud rate → significant class imbalance
- transaction_id is a pure identifier — will be dropped before modeling
- Boolean columns (is_foreign_transaction, is_new_merchant, used_vpn, etc.) are already 0/1-ready
- Categorical columns (merchant_category, card_type, auth_method, channel, device_type) will need encoding
- No cleaning of missing/duplicate data required — can move straight to EDA'''
```
