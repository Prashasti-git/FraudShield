```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

df = pd.read_csv('../data/raw/credit_card_fraud_2026.csv')
df.shape
```

```python
print('Missing values:', df.isnull().sum().sum())
print('Duplicate rows:', df.duplicated().sum())
```

```python
#Identify which columns to check for outliers
outlier_cols = ['amount_usd', 'account_balance_usd', 'distance_from_home_km',
                 'hours_since_last_txn', 'txn_count_last_24h', 'velocity_score',
                 'cvv_retry_count', 'merchant_risk_score', 'prior_disputes',
                 'card_age_months', 'customer_age']
```

```python
#Visualize outliers with boxplots
fig, axes = plt.subplots(3, 4, figsize=(18, 12))
for ax, col in zip(axes.ravel(), outlier_cols):
    df[col].plot(kind='box', ax=ax)
    ax.set_title(col, fontsize=10)
plt.suptitle('Boxplots — Outlier Check Across Features', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
#Quantify outliers using IQR
for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f'{col}: {len(outliers)} outliers ({len(outliers)/len(df)*100:.2f}%)')
```

```python
#check whether "outliers" are actually fraud
for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    mask = (df[col] < lower) | (df[col] > upper)
    if mask.sum() > 0:
        fraud_rate_outliers = df.loc[mask, 'is_fraud'].mean()
        fraud_rate_overall = df['is_fraud'].mean()
        print(f'{col}: fraud rate in outliers = {fraud_rate_outliers:.4f} | overall = {fraud_rate_overall:.4f}')
```

```python
## Outlier Policy

'''QR-based outlier detection was run across all numeric features, and fraud rate
within outliers was compared to the overall fraud rate (1.69%) for each column.
Outliers in velocity_score, merchant_risk_score, cvv_retry_count, and 
txn_count_last_24h showed 3-6x higher fraud rates than the dataset average, 
confirming these are genuine fraud cases rather than data errors. No column 
showed outliers behaving as pure noise. Decision: no outlier removal performed
all extreme values are retained for modeling.'''
```
