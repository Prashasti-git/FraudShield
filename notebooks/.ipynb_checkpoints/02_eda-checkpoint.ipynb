```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/raw/credit_card_fraud_2026.csv')
df.shape
```

```python
##Univariate analysis to look at the target and key features on their own
```

```python
##target distribution
plt.figure(figsize=(6, 4))
df['is_fraud'].value_counts().plot(kind='bar', color=['lightblue', 'tomato'])
plt.title('Fraud vs Normal Transaction Count')
plt.xlabel('is_fraud')
plt.ylabel('Count')
plt.show()
```

```python
##distribution of transaction amount
plt.figure(figsize=(8, 4))
sns.histplot(df['amount_usd'], bins=50, kde=True, color='grey')
plt.title('Distribution of Transaction Amount')
plt.show()
```

```python
##distribution of a few key numeric features at once
num_cols = ['amount_usd', 'account_balance_usd', 'distance_from_home_km',
            'velocity_score', 'merchant_risk_score', 'cvv_retry_count']

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, col in zip(axes.ravel(), num_cols):
    df[col].plot(kind='box', ax=ax)
    ax.set_title(col, fontsize=10)
plt.suptitle('Boxplots for Outlier Check', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
##Bivariate analysis to compare fraud vs normal on key features
```

```python
##amount by fraud label
plt.figure(figsize=(8, 5))
sns.boxplot(x='is_fraud', y='amount_usd', hue='is_fraud', data=df,
            palette={0: 'steelblue', 1: 'tomato'}, legend=False)
plt.title('Transaction Amount by Fraud Label')
plt.show()
```

```python
##the risk flag columns vs fraud rate
risk_cols = ['is_foreign_transaction', 'ip_country_mismatch',
             'billing_shipping_mismatch', 'is_ai_generated_scam_attempt', 'used_vpn']

for col in risk_cols:
    print(df.groupby(col)['is_fraud'].mean())
    print()
```

```python
'''This prints the fraud rate for each value of each flaga and this is exactly how our original presentation found that certain flags (like Is_International_Transaction) correlate much more strongly with fraud than others. we also need to watch for any flag where the fraud rate is noticeably higher when the flag is True.'''
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='is_fraud', y='velocity_score', hue='is_fraud', data=df,
            palette={0: 'steelblue', 1: 'tomato'}, legend=False)
plt.title('Velocity Score by Fraud Label')
plt.show()
```

```python
##Multivariate correlation heatmap
```

```python
numeric_df = df.select_dtypes(include=[np.number])
plt.figure(figsize=(12, 9))
sns.heatmap(numeric_df.corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap — Numeric Features')
plt.tight_layout()
plt.show()
```

```python
'''if we look specifically at the is_fraud row/column here — whichever features have the strongest (positive or negative) correlation with it are our early candidates for "meaningful predictors," the same kind of insight our original presentation highlighted for Is_International_Transaction and Unusual_Time_Transaction.'''
```

```python
##Categorical features vs fraud
cat_cols = ['merchant_category', 'card_type', 'auth_method', 'channel', 'device_type']

for col in cat_cols:
    print(df.groupby(col)['is_fraud'].mean().sort_values(ascending=False))
    print()
```
